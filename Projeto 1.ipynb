```python
# --- FUNÇÕES EM PYTHON: GESTÃO DE NOTAS DE ALUNOS ---

# 1. FUNÇÃO PARA CADASTRAR AS NOTAS
# Esta função é responsável por pedir as notas ao usuário e guardá-las em uma lista.
def cadastrar_notas():
    """Pede ao usuário para inserir notas até que 'sair' seja digitado.
    Retorna uma lista contendo as notas."""

    # Criamos uma lista vazia para armazenar as notas que o usuário digitar.
    notas_aluno = []
    print("--- Cadastro de Notas ---")
    print("Digite as notas do aluno. Digite 'sair' para finalizar.")

    # Usamos um loop infinito para que o usuário possa digitar quantas notas quiser.
    while True:
        # Pede para o usuário digitar uma nota.
        entrada = input(f"Digite a nota {len(notas_aluno) + 1} (ou 'sair'): ")

        # Se o usuário digitar 'sair', o loop é interrompido com o comando 'break'.
        if entrada.lower() == 'sair':
            break

        # Usamos try/except para garantir que o programa não quebre se o usuário digitar
        # algo que não seja um número.
        try:
            # Tenta converter o texto digitado para um número (float permite casas decimais).
            nota = float(entrada)
            # Adiciona a nota convertida à nossa lista de notas.
            notas_aluno.append(nota)
        except ValueError:
            # Se a conversão falhar, avisa o usuário e continua o loop.
            print("Entrada inválida. Por favor, digite um número ou 'sair'.")

    return notas_aluno

# 2. FUNÇÃO PARA CALCULAR A MÉDIA
# Esta função recebe a lista de notas e calcula a média.
def calcular_media(notas):
    """Calcula a média de uma lista de notas.
    Retorna a média ou 0 se a lista estiver vazia."""

    # Verifica se a lista de notas não está vazia para evitar erro de divisão por zero.
    if len(notas) > 0:
        # Soma todos os itens da lista e divide pelo número de itens.
        media = sum(notas) / len(notas)
        return media
    else:
        # Se a lista estiver vazia, a média é 0.
        return 0

# 3. FUNÇÃO PARA DETERMINAR A SITUAÇÃO
# Esta função verifica a média e retorna se o aluno foi aprovado ou reprovado.
def determinar_situacao(media):
    """Determina a situação do aluno com base na média.
    Retorna 'Aprovado' ou 'Reprovado'."""

    # Estrutura condicional (if/else) para verificar a regra de aprovação.
    if media >= 7.0:
        return "Aprovado" # Se a média for maior ou igual a 7.
    else:
        return "Reprovado" # Se a média for menor que 7.

# 4. FUNÇÃO PARA EXIBIR O RELATÓRIO FINAL
# Esta função organiza e mostra todas as informações na tela.
def exibir_relatorio(notas, media, situacao):
    """Exibe o relatório final com as notas, média e situação."""

    print("\n--- Relatório Final do Aluno ---")
    # Mostra todas as notas que foram inseridas.
    print(f"Notas inseridas: {notas}")
    # Mostra a média calculada, formatada com 2 casas decimais.
    print(f"Média final: {media:.2f}")
    # Mostra a situação final do aluno.
    print(f"Situação: {situacao}")
    print("---------------------------------")


# --- EXECUÇÃO PRINCIPAL DO PROGRAMA ---
# O código abaixo é o que "orquestra" as funções, chamando uma após a outra.

# Chama a função para cadastrar as notas e guarda o resultado na variável 'lista_de_notas'.
lista_de_notas = cadastrar_notas()

# Chama a função para calcular a média, usando a lista criada acima.
media_final = calcular_media(lista_de_notas)

# Chama a função para determinar a situação, usando a média calculada.
situacao_aluno = determinar_situacao(media_final)

# Finalmente, chama a função que exibe o relatório completo com todas as informações.
exibir_relatorio(lista_de_notas, media_final, situacao_aluno)
```
